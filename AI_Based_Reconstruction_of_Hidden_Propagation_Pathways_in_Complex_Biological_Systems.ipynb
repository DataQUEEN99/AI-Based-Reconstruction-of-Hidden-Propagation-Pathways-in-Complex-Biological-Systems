```python
# ================================================================
# CAUSAL-ECHO AI
# Reconstructing Hidden Propagation Pathways in Complex Systems
#
# COMPLETE ONE-CELL GOOGLE COLAB RESEARCH PROTOTYPE
# ================================================================

!pip -q install numpy pandas matplotlib seaborn scikit-learn networkx scipy xgboost shap

# ================================================================
# 1. IMPORTS
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx

from itertools import combinations

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.ensemble import (
    RandomForestRegressor,
    RandomForestClassifier,
    IsolationForest
)
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    silhouette_score
)
from sklearn.inspection import permutation_importance

from scipy.stats import pearsonr, spearmanr

from xgboost import XGBRegressor, XGBClassifier

import shap
import warnings

warnings.filterwarnings("ignore")

np.random.seed(42)

print("="*90)
print("CAUSAL-ECHO AI")
print("AI-Based Reconstruction of Hidden Propagation Pathways")
print("="*90)

# ================================================================
# 2. PROJECT CONFIGURATION
# ================================================================

N_SAMPLES = 2500
N_TIMEPOINTS = 10
N_BIO_FEATURES = 25
NOISE_LEVEL = 0.08

feature_names = [
    "Gene_Expression_A",
    "Gene_Expression_B",
    "Gene_Expression_C",
    "Gene_Expression_D",
    "Gene_Expression_E",
    "Protein_Activity_A",
    "Protein_Activity_B",
    "Protein_Activity_C",
    "Metabolic_Activity",
    "ATP_Level",
    "ROS_Level",
    "Oxidative_Stress",
    "Inflammation",
    "DNA_Damage",
    "DNA_Repair",
    "Mitochondrial_Activity",
    "Protein_Synthesis",
    "Autophagy",
    "Cell_Cycle",
    "Apoptosis_Signal",
    "Immune_Signaling",
    "Growth_Signaling",
    "Membrane_Integrity",
    "Cellular_Stress",
    "Final_Cell_State"
]

print("\nConfiguration")
print("-"*60)
print("Samples:", N_SAMPLES)
print("Timepoints:", N_TIMEPOINTS)
print("Biological variables:", N_BIO_FEATURES)
print("Noise level:", NOISE_LEVEL)

# ================================================================
# 3. HIDDEN BIOLOGICAL CAUSAL STRUCTURE
# ================================================================
#
# IMPORTANT:
# This structure is hidden from the ML discovery algorithms.
# It is used only to generate synthetic ground truth.
#
# Example:
#
# Initial_Perturbation
#        ↓
# Gene A
#        ↓
# Gene B
#        ↓
# Protein A
#        ↓
# ROS
#        ↓
# Oxidative Stress
#        ↓
# Mitochondrial Activity
#        ↓
# ATP
#        ↓
# Cellular Stress
#        ↓
# Final Cell State
#
# ================================================================

true_edges = [

    ("Gene_Expression_A", "Gene_Expression_B"),
    ("Gene_Expression_B", "Gene_Expression_C"),
    ("Gene_Expression_C", "Protein_Activity_A"),

    ("Protein_Activity_A", "ROS_Level"),
    ("ROS_Level", "Oxidative_Stress"),

    ("Oxidative_Stress", "Mitochondrial_Activity"),
    ("Mitochondrial_Activity", "ATP_Level"),

    ("ATP_Level", "Cellular_Stress"),

    ("Oxidative_Stress", "DNA_Damage"),
    ("DNA_Damage", "DNA_Repair"),

    ("DNA_Damage", "Apoptosis_Signal"),

    ("Gene_Expression_D", "Inflammation"),
    ("Inflammation", "Immune_Signaling"),

    ("Gene_Expression_E", "Growth_Signaling"),

    ("Protein_Activity_B", "Protein_Synthesis"),
    ("Protein_Synthesis", "Protein_Quality_Control")
    if "Protein_Quality_Control" in feature_names
    else ("Protein_Synthesis", "Autophagy"),

    ("ATP_Level", "Cell_Cycle"),
    ("Cell_Cycle", "Apoptosis_Signal"),

    ("Autophagy", "Cellular_Stress"),
    ("Immune_Signaling", "Cellular_Stress"),

    ("Cellular_Stress", "Membrane_Integrity"),
    ("Membrane_Integrity", "Final_Cell_State"),

    ("Apoptosis_Signal", "Final_Cell_State"),
    ("Growth_Signaling", "Final_Cell_State"),
    ("ATP_Level", "Final_Cell_State")
]

# Remove invalid self/nonexistent edges
true_edges = [
    (a,b)
    for a,b in true_edges
    if a in feature_names and b in feature_names and a != b
]

true_edge_set = set(true_edges)

print("\nHidden causal structure contains:")
print(len(true_edges), "ground-truth edges")

# ================================================================
# 4. CREATE DIRECTED GROUND-TRUTH GRAPH
# ================================================================

true_graph = nx.DiGraph()

true_graph.add_nodes_from(
    feature_names
)

true_graph.add_edges_from(
    true_edges
)

print(
    "Ground-truth nodes:",
    true_graph.number_of_nodes()
)

print(
    "Ground-truth edges:",
    true_graph.number_of_edges()
)

# ================================================================
# 5. GENERATE TEMPORAL BIOLOGICAL DATA
# ================================================================
#
# Each sample has a perturbation intensity.
# The hidden causal chain propagates across time.
#
# ================================================================

records = []

for sample_id in range(N_SAMPLES):

    # ------------------------------------------------------------
    # Initial perturbation
    # ------------------------------------------------------------

    perturbation = np.random.normal(
        0,
        1
    )

    baseline = np.random.normal(
        0,
        0.15,
        N_BIO_FEATURES
    )

    values = {}

    # ------------------------------------------------------------
    # Time-dependent perturbation
    # ------------------------------------------------------------

    for t in range(N_TIMEPOINTS):

        progression = (
            perturbation *
            (t / (N_TIMEPOINTS - 1))
        )

        # --------------------------------------------------------
        # Root biological variables
        # --------------------------------------------------------

        A = (
            baseline[0]
            + progression * 1.4
            + np.random.normal(0, NOISE_LEVEL)
        )

        B = (
            baseline[1]
            + 0.70*A
            + progression*0.15
            + np.random.normal(0, NOISE_LEVEL)
        )

        C = (
            baseline[2]
            + 0.75*B
            + np.random.normal(0, NOISE_LEVEL)
        )

        D = (
            baseline[3]
            + np.random.normal(0, NOISE_LEVEL)
        )

        E = (
            baseline[4]
            + np.random.normal(0, NOISE_LEVEL)
        )

        Protein_A = (
            baseline[5]
            + 0.85*C
            + np.random.normal(0, NOISE_LEVEL)
        )

        Protein_B = (
            baseline[6]
            + np.random.normal(0, NOISE_LEVEL)
        )

        Protein_C = (
            baseline[7]
            + np.random.normal(0, NOISE_LEVEL)
        )

        # --------------------------------------------------------
        # Metabolic pathway
        # --------------------------------------------------------

        ROS = (
            baseline[10]
            + 0.80*Protein_A
            + 0.25*progression
            + np.random.normal(0, NOISE_LEVEL)
        )

        OxStress = (
            baseline[11]
            + 0.85*ROS
            + np.random.normal(0, NOISE_LEVEL)
        )

        Mito = (
            baseline[15]
            - 0.70*OxStress
            + np.random.normal(0, NOISE_LEVEL)
        )

        ATP = (
            baseline[9]
            + 0.80*Mito
            - 0.25*OxStress
            + np.random.normal(0, NOISE_LEVEL)
        )

        # --------------------------------------------------------
        # DNA pathway
        # --------------------------------------------------------

        DNA_damage = (
            baseline[13]
            + 0.75*OxStress
            + np.random.normal(0, NOISE_LEVEL)
        )

        DNA_repair = (
            baseline[14]
            + 0.65*DNA_damage
            + np.random.normal(0, NOISE_LEVEL)
        )

        # --------------------------------------------------------
        # Immune pathway
        # --------------------------------------------------------

        Inflammation = (
            baseline[12]
            + 0.70*D
            + 0.20*progression
            + np.random.normal(0, NOISE_LEVEL)
        )

        Immune = (
            baseline[20]
            + 0.75*Inflammation
            + np.random.normal(0, NOISE_LEVEL)
        )

        # --------------------------------------------------------
        # Growth pathway
        # --------------------------------------------------------

        Growth = (
            baseline[21]
            + 0.80*E
            + np.random.normal(0, NOISE_LEVEL)
        )

        # --------------------------------------------------------
        # Protein / cellular maintenance
        # --------------------------------------------------------

        Protein_Synthesis = (
            baseline[16]
            + 0.60*Protein_B
            + 0.25*ATP
            + np.random.normal(0, NOISE_LEVEL)
        )

        Autophagy = (
            baseline[17]
            + 0.55*OxStress
            - 0.25*ATP
            + np.random.normal(0, NOISE_LEVEL)
        )

        Cell_Cycle = (
            baseline[18]
            + 0.55*ATP
            + 0.30*Growth
            + np.random.normal(0, NOISE_LEVEL)
        )

        Apoptosis = (
            baseline[19]
            + 0.70*DNA_damage
            + 0.50*OxStress
            - 0.30*ATP
            + np.random.normal(0, NOISE_LEVEL)
        )

        Membrane = (
            baseline[22]
            + 0.55*ATP
            - 0.60*OxStress
            - 0.50*Apoptosis
            + np.random.normal(0, NOISE_LEVEL)
        )

        CellularStress = (
            baseline[23]
            + 0.55*OxStress
            + 0.50*Inflammation
            + 0.40*Autophagy
            - 0.40*ATP
            + np.random.normal(0, NOISE_LEVEL)
        )

        FinalState = (
            0.45*CellularStress
            + 0.35*Apoptosis
            + 0.30*Growth
            + 0.25*ATP
            + np.random.normal(0, NOISE_LEVEL)
        )

        row = {

            "sample_id":
                sample_id,

            "time":
                t,

            "perturbation":
                perturbation,

            "Gene_Expression_A":
                A,

            "Gene_Expression_B":
                B,

            "Gene_Expression_C":
                C,

            "Gene_Expression_D":
                D,

            "Gene_Expression_E":
                E,

            "Protein_Activity_A":
                Protein_A,

            "Protein_Activity_B":
                Protein_B,

            "Protein_Activity_C":
                Protein_C,

            "Metabolic_Activity":
                0.6*ATP + np.random.normal(0, NOISE_LEVEL),

            "ATP_Level":
                ATP,

            "ROS_Level":
                ROS,

            "Oxidative_Stress":
                OxStress,

            "Inflammation":
                Inflammation,

            "DNA_Damage":
                DNA_damage,

            "DNA_Repair":
                DNA_repair,

            "Mitochondrial_Activity":
                Mito,

            "Protein_Synthesis":
                Protein_Synthesis,

            "Autophagy":
                Autophagy,

            "Cell_Cycle":
                Cell_Cycle,

            "Apoptosis_Signal":
                Apoptosis,

            "Immune_Signaling":
                Immune,

            "Growth_Signaling":
                Growth,

            "Membrane_Integrity":
                Membrane,

            "Cellular_Stress":
                CellularStress,

            "Final_Cell_State":
                FinalState
        }

        records.append(row)

data = pd.DataFrame(
    records
)

print("\nGenerated temporal biological dataset:")
print(data.shape)

display(
    data.head()
)

# ================================================================
# 6. DATA QUALITY ANALYSIS
# ================================================================

print("\nDATA QUALITY")
print("-"*60)

print(
    "Missing values:",
    data.isna().sum().sum()
)

print(
    "Duplicate rows:",
    data.duplicated().sum()
)

print(
    "Samples:",
    data["sample_id"].nunique()
)

print(
    "Timepoints:",
    data["time"].nunique()
)

# ================================================================
# 7. BASIC STATISTICAL SUMMARY
# ================================================================

summary = data[
    feature_names
].describe().T

summary[
    "variance"
] = data[
    feature_names
].var()

display(
    summary.head(15)
)

# ================================================================
# 8. CORRELATION NETWORK
# ================================================================

corr = data[
    feature_names
].corr(
    method="spearman"
)

plt.figure(
    figsize=(18,14)
)

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0,
    square=True
)

plt.title(
    "Biological Feature Correlation Matrix"
)

plt.tight_layout()
plt.show()

# ================================================================
# 9. TEMPORAL FEATURE TRAJECTORIES
# ================================================================

important_features = [

    "Gene_Expression_A",
    "Gene_Expression_B",
    "Protein_Activity_A",
    "ROS_Level",
    "Oxidative_Stress",
    "ATP_Level",
    "DNA_Damage",
    "Cellular_Stress",
    "Final_Cell_State"
]

plt.figure(
    figsize=(15,8)
)

for feature in important_features:

    temporal_mean = (
        data
        .groupby("time")[feature]
        .mean()
    )

    plt.plot(
        temporal_mean.index,
        temporal_mean.values,
        marker="o",
        linewidth=2,
        label=feature
    )

plt.title(
    "Temporal Biological Cascade"
)

plt.xlabel(
    "Time"
)

plt.ylabel(
    "Mean standardized biological signal"
)

plt.legend(
    bbox_to_anchor=(1.02,1),
    loc="upper left"
)

plt.grid(
    alpha=0.25
)

plt.tight_layout()
plt.show()

# ================================================================
# 10. TEMPORAL PRECEDENCE ANALYSIS
# ================================================================
#
# A simple computational heuristic:
# measure when each feature starts deviating strongly
# from its baseline.
#
# This is NOT proof of causality.
#
# ================================================================

temporal_records = []

for feature in feature_names:

    means = (
        data
        .groupby("time")[feature]
        .mean()
        .reindex(
            range(N_TIMEPOINTS)
        )
    )

    baseline = means.iloc[0]

    deviations = (
        abs(means - baseline)
    )

    max_deviation = deviations.max()

    threshold = (
        0.30 *
        max_deviation
    )

    candidate_times = (
        deviations[
            deviations >= threshold
        ]
        .index
    )

    first_change = (
        candidate_times[0]
        if len(candidate_times) > 0
        else N_TIMEPOINTS
    )

    temporal_records.append({

        "feature":
            feature,

        "first_change_time":
            first_change,

        "maximum_deviation":
            max_deviation,

        "final_change":
            means.iloc[-1] - baseline
    })

temporal_df = pd.DataFrame(
    temporal_records
)

temporal_df = (
    temporal_df
    .sort_values(
        "first_change_time"
    )
    .reset_index(drop=True)
)

print("\nTemporal precedence analysis")

display(
    temporal_df
)

# ================================================================
# 11. MACHINE LEARNING — RANDOM FOREST REGRESSION
# ================================================================
#
# Predict Final_Cell_State using all other biological features.
#
# ================================================================

target = "Final_Cell_State"

predictor_features = [
    f for f in feature_names
    if f != target
]

X = data[
    predictor_features
]

y = data[
    target
]

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42
)

rf = RandomForestRegressor(

    n_estimators=250,

    max_depth=12,

    min_samples_leaf=3,

    random_state=42,

    n_jobs=-1
)

rf.fit(
    X_train,
    y_train
)

rf_pred = rf.predict(
    X_test
)

rf_mae = mean_absolute_error(
    y_test,
    rf_pred
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_pred
    )
)

rf_r2 = r2_score(
    y_test,
    rf_pred
)

print("\nRandom Forest")

print(
    "MAE:",
    round(rf_mae,4)
)

print(
    "RMSE:",
    round(rf_rmse,4)
)

print(
    "R²:",
    round(rf_r2,4)
)

# ================================================================
# 12. XGBOOST MODEL
# ================================================================

xgb = XGBRegressor(

    n_estimators=300,

    max_depth=5,

    learning_rate=0.04,

    subsample=0.85,

    colsample_bytree=0.85,

    objective="reg:squarederror",

    random_state=42,

    n_jobs=-1
)

xgb.fit(
    X_train,
    y_train
)

xgb_pred = xgb.predict(
    X_test
)

xgb_mae = mean_absolute_error(
    y_test,
    xgb_pred
)

xgb_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        xgb_pred
    )
)

xgb_r2 = r2_score(
    y_test,
    xgb_pred
)

print("\nXGBoost")

print(
    "MAE:",
    round(xgb_mae,4)
)

print(
    "RMSE:",
    round(xgb_rmse,4)
)

print(
    "R²:",
    round(xgb_r2,4)
)

# ================================================================
# 13. MODEL COMPARISON
# ================================================================

model_results = pd.DataFrame({

    "model": [
        "Random Forest",
        "XGBoost"
    ],

    "MAE": [
        rf_mae,
        xgb_mae
    ],

    "RMSE": [
        rf_rmse,
        xgb_rmse
    ],

    "R2": [
        rf_r2,
        xgb_r2
    ]
})

display(
    model_results
)

# ================================================================
# 14. ACTUAL VS PREDICTED
# ================================================================

plt.figure(
    figsize=(9,7)
)

plt.scatter(
    y_test,
    xgb_pred,
    alpha=0.5
)

minimum = min(
    y_test.min(),
    xgb_pred.min()
)

maximum = max(
    y_test.max(),
    xgb_pred.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.xlabel(
    "Actual Final Cell State"
)

plt.ylabel(
    "Predicted Final Cell State"
)

plt.title(
    "XGBoost — Actual vs Predicted"
)

plt.grid(
    alpha=0.25
)

plt.show()

# ================================================================
# 15. FEATURE IMPORTANCE
# ================================================================

importance_df = pd.DataFrame({

    "feature":
        predictor_features,

    "importance":
        xgb.feature_importances_
})

importance_df = (
    importance_df
    .sort_values(
        "importance",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nXGBoost Feature Importance")

display(
    importance_df.head(20)
)

plt.figure(
    figsize=(12,8)
)

top_imp = (
    importance_df
    .head(15)
    .sort_values(
        "importance"
    )
)

plt.barh(
    top_imp["feature"],
    top_imp["importance"]
)

plt.title(
    "Top Features Driving Final Cell-State Prediction"
)

plt.xlabel(
    "XGBoost Importance"
)

plt.tight_layout()
plt.show()

# ================================================================
# 16. PERMUTATION IMPORTANCE
# ================================================================

perm = permutation_importance(

    xgb,

    X_test,

    y_test,

    n_repeats=5,

    random_state=42,

    scoring="r2"
)

permutation_df = pd.DataFrame({

    "feature":
        predictor_features,

    "permutation_importance":
        perm.importances_mean
})

permutation_df = (
    permutation_df
    .sort_values(
        "permutation_importance",
        ascending=False
    )
)

print("\nPermutation importance")

display(
    permutation_df.head(15)
)

# ================================================================
# 17. SHAP EXPLAINABILITY
# ================================================================

print("\nCalculating SHAP explanations...")

explainer = shap.TreeExplainer(
    xgb
)

sample_for_shap = (
    X_test
    .sample(
        min(500, len(X_test)),
        random_state=42
    )
)

shap_values = explainer(
    sample_for_shap
)

plt.figure(
    figsize=(12,8)
)

shap.summary_plot(
    shap_values.values,
    sample_for_shap,
    show=False
)

plt.title(
    "SHAP — Global Model Explanation"
)

plt.tight_layout()
plt.show()

# ================================================================
# 18. SHAP IMPORTANCE TABLE
# ================================================================

mean_abs_shap = np.abs(
    shap_values.values
).mean(
    axis=0
)

shap_df = pd.DataFrame({

    "feature":
        predictor_features,

    "mean_abs_SHAP":
        mean_abs_shap
})

shap_df = (
    shap_df
    .sort_values(
        "mean_abs_SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nSHAP Feature Importance")

display(
    shap_df.head(20)
)

# ================================================================
# 19. ANOMALY DETECTION
# ================================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    data[
        predictor_features
    ]
)

isolation = IsolationForest(

    n_estimators=250,

    contamination=0.05,

    random_state=42
)

anomaly_labels = isolation.fit_predict(
    X_scaled
)

anomaly_scores = -isolation.score_samples(
    X_scaled
)

data["anomaly_label"] = anomaly_labels

data["anomaly_score"] = anomaly_scores

print("\nAnomaly detection")

print(
    "Detected anomalies:",
    (anomaly_labels == -1).sum()
)

# ================================================================
# 20. ANOMALY SCORE THROUGH TIME
# ================================================================

anomaly_temporal = (
    data
    .groupby("time")["anomaly_score"]
    .mean()
)

plt.figure(
    figsize=(11,6)
)

plt.plot(
    anomaly_temporal.index,
    anomaly_temporal.values,
    marker="o",
    linewidth=3
)

plt.title(
    "System Anomaly Intensity Through Time"
)

plt.xlabel(
    "Time"
)

plt.ylabel(
    "Mean anomaly score"
)

plt.grid(
    alpha=0.3
)

plt.show()

# ================================================================
# 21. HIDDEN STATE DISCOVERY
# ================================================================

state_model = KMeans(

    n_clusters=4,

    random_state=42,

    n_init=20
)

state_labels = state_model.fit_predict(
    X_scaled
)

data["hidden_state"] = state_labels

state_profile = (
    data
    .groupby("hidden_state")[
        predictor_features
    ]
    .mean()
)

print("\nHidden biological states")

display(
    state_profile
)

# ================================================================
# 22. PCA STATE VISUALIZATION
# ================================================================

pca = PCA(
    n_components=2,
    random_state=42
)

pca_data = pca.fit_transform(
    X_scaled
)

plt.figure(
    figsize=(11,8)
)

plt.scatter(

    pca_data[:,0],

    pca_data[:,1],

    c=data["hidden_state"],

    cmap="tab10",

    alpha=0.55,

    s=20
)

plt.xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)"
)

plt.ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)"
)

plt.title(
    "Hidden Biological State Landscape"
)

plt.grid(
    alpha=0.25
)

plt.show()

# ================================================================
# 23. TEMPORAL STATE TRANSITION MATRIX
# ================================================================

state_transition_records = []

for sample_id in data[
    "sample_id"
].unique():

    sample = (
        data[
            data["sample_id"] == sample_id
        ]
        .sort_values("time")
    )

    states = sample[
        "hidden_state"
    ].values

    for i in range(
        len(states)-1
    ):

        state_transition_records.append({

            "sample_id":
                sample_id,

            "time":
                i,

            "from_state":
                states[i],

            "to_state":
                states[i+1]
        })

transitions_df = pd.DataFrame(
    state_transition_records
)

transition_matrix = pd.crosstab(

    transitions_df[
        "from_state"
    ],

    transitions_df[
        "to_state"
    ]
)

print("\nHidden-state transition matrix")

display(
    transition_matrix
)

plt.figure(
    figsize=(8,7)
)

sns.heatmap(
    transition_matrix,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Biological State Transition Matrix"
)

plt.xlabel(
    "To State"
)

plt.ylabel(
    "From State"
)

plt.show()

# ================================================================
# 24. LAGGED RELATIONSHIP ANALYSIS
# ================================================================
#
# Check whether changes in feature A at time t are associated
# with feature B at time t+1.
#
# This provides temporal evidence but does NOT prove causality.
#
# ================================================================

lag_records = []

for source, target_feature in combinations(
    feature_names,
    2
):

    source_values = []
    target_values = []

    for sample_id in data[
        "sample_id"
    ].unique():

        sample = (
            data[
                data["sample_id"] == sample_id
            ]
            .sort_values("time")
        )

        if len(sample) < 2:
            continue

        source_values.extend(
            sample[source]
            .iloc[:-1]
            .values
        )

        target_values.extend(
            sample[target_feature]
            .iloc[1:]
            .values
        )

    if len(source_values) > 10:

        correlation, pvalue = pearsonr(
            source_values,
            target_values
        )

        lag_records.append({

            "source":
                source,

            "target":
                target_feature,

            "lag_correlation":
                correlation,

            "p_value":
                pvalue
        })

lag_df = pd.DataFrame(
    lag_records
)

lag_df["abs_lag_correlation"] = (
    abs(
        lag_df[
            "lag_correlation"
        ]
    )
)

lag_df = (
    lag_df
    .sort_values(
        "abs_lag_correlation",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nStrongest lagged relationships")

display(
    lag_df.head(25)
)

# ================================================================
# 25. DISCOVERED PROPAGATION GRAPH
# ================================================================

DISCOVERY_THRESHOLD = 0.35

discovered_graph = nx.DiGraph()

discovered_graph.add_nodes_from(
    feature_names
)

for _, row in lag_df.iterrows():

    if (
        abs(
            row["lag_correlation"]
        )
        >= DISCOVERY_THRESHOLD
    ):

        discovered_graph.add_edge(

            row["source"],

            row["target"],

            weight=abs(
                row["lag_correlation"]
            ),

            signed_weight=
                row["lag_correlation"]
        )

print("\nDiscovered propagation network")

print(
    "Nodes:",
    discovered_graph.number_of_nodes()
)

print(
    "Edges:",
    discovered_graph.number_of_edges()
)

# ================================================================
# 26. COMPARE DISCOVERED EDGES WITH HIDDEN TRUTH
# ================================================================

discovered_edges = set(
    discovered_graph.edges()
)

true_edges_found = (
    discovered_edges
    .intersection(
        true_edge_set
    )
)

false_edges = (
    discovered_edges
    - true_edge_set
)

missed_edges = (
    true_edge_set
    - discovered_edges
)

precision_discovery = (
    len(true_edges_found) /
    max(len(discovered_edges),1)
)

recall_discovery = (
    len(true_edges_found) /
    max(len(true_edge_set),1)
)

print("\nHidden-structure recovery")

print(
    "True edges:",
    len(true_edge_set)
)

print(
    "Discovered edges:",
    len(discovered_edges)
)

print(
    "Correctly recovered:",
    len(true_edges_found)
)

print(
    "False edges:",
    len(false_edges)
)

print(
    "Missed true edges:",
    len(missed_edges)
)

print(
    "Structural precision:",
    round(
        precision_discovery,
        4
    )
)

print(
    "Structural recall:",
    round(
        recall_discovery,
        4
    )
)

# ================================================================
# 27. DISCOVERED NETWORK VISUALIZATION
# ================================================================

plt.figure(
    figsize=(16,12)
)

pos = nx.spring_layout(
    discovered_graph,
    seed=42,
    k=0.55
)

nx.draw_networkx_nodes(

    discovered_graph,

    pos,

    node_size=900,

    alpha=0.85
)

nx.draw_networkx_edges(

    discovered_graph,

    pos,

    arrows=True,

    arrowsize=15,

    alpha=0.25,

    width=1.5,

    connectionstyle="arc3,rad=0.05"
)

nx.draw_networkx_labels(

    discovered_graph,

    pos,

    font_size=8
)

plt.title(
    "CAUSAL-ECHO — Discovered Temporal Propagation Network",
    fontsize=17
)

plt.axis("off")
plt.show()

# ================================================================
# 28. EDGE-LEVEL EVIDENCE TABLE
# ================================================================

edge_records = []

for source, target in discovered_graph.edges():

    lag_value = lag_df[
        (
            lag_df["source"] == source
        )
        &
        (
            lag_df["target"] == target
        )
    ]

    if len(lag_value) == 0:
        continue

    correlation_value = (
        lag_value.iloc[0]
        ["lag_correlation"]
    )

    is_true = (
        (source, target)
        in true_edge_set
    )

    edge_records.append({

        "source":
            source,

        "target":
            target,

        "lag_correlation":
            correlation_value,

        "absolute_strength":
            abs(correlation_value),

        "ground_truth_edge":
            is_true
    })

edge_evidence_df = pd.DataFrame(
    edge_records
)

edge_evidence_df = (
    edge_evidence_df
    .sort_values(
        "absolute_strength",
        ascending=False
    )
)

display(
    edge_evidence_df.head(30)
)

# ================================================================
# 29. NETWORK CENTRALITY
# ================================================================

if discovered_graph.number_of_edges() > 0:

    in_degree = dict(
        discovered_graph.in_degree()
    )

    out_degree = dict(
        discovered_graph.out_degree()
    )

    centrality = nx.betweenness_centrality(
        discovered_graph
    )

else:

    in_degree = {
        node:0
        for node in feature_names
    }

    out_degree = {
        node:0
        for node in feature_names
    }

    centrality = {
        node:0
        for node in feature_names
    }

network_df = pd.DataFrame({

    "feature":
        feature_names,

    "in_degree":
        [
            in_degree.get(
                f,
                0
            )
            for f in feature_names
        ],

    "out_degree":
        [
            out_degree.get(
                f,
                0
            )
            for f in feature_names
        ],

    "betweenness":
        [
            centrality.get(
                f,
                0
            )
            for f in feature_names
        ]
})

print("\nNetwork-central features")

display(
    network_df
    .sort_values(
        "betweenness",
        ascending=False
    )
    .head(15)
)

# ================================================================
# 30. TEMPORAL + NETWORK + ML SCORE
# ================================================================
#
# Causal-Echo Score is a research prioritization heuristic.
# It does NOT establish causality.
#
# ================================================================

merged_score = (
    temporal_df[
        [
            "feature",
            "first_change_time",
            "maximum_deviation",
            "final_change"
        ]
    ]
    .merge(
        network_df,
        on="feature",
        how="left"
    )
)

merged_score = (
    merged_score
    .merge(
        shap_df.rename(
            columns={
                "feature":
                    "feature"
            }
        ),
        on="feature",
        how="left"
    )
)

# Earlier change = higher temporal score
merged_score[
    "temporal_score"
] = (

    1 -

    (
        merged_score[
            "first_change_time"
        ]
        /
        max(
            N_TIMEPOINTS,
            1
        )
    )
)

merged_score[
    "temporal_score"
] = (
    merged_score[
        "temporal_score"
    ]
    .clip(0,1)
)

# Normalize network score
def minmax(series):

    minimum = series.min()
    maximum = series.max()

    if maximum == minimum:
        return pd.Series(
            0.5,
            index=series.index
        )

    return (
        (series - minimum)
        /
        (maximum - minimum)
    )

merged_score[
    "network_score"
] = minmax(
    merged_score[
        "betweenness"
    ]
)

merged_score[
    "SHAP_score"
] = minmax(
    merged_score[
        "mean_abs_SHAP"
    ]
)

merged_score[
    "deviation_score"
] = minmax(
    merged_score[
        "maximum_deviation"
    ]
)

# ---------------------------------------------------------------
# Causal-Echo Score
# ---------------------------------------------------------------

merged_score[
    "Causal_Echo_Score"
] = (

    0.30 *
    merged_score[
        "temporal_score"
    ]

    +

    0.25 *
    merged_score[
        "network_score"
    ]

    +

    0.25 *
    merged_score[
        "SHAP_score"
    ]

    +

    0.20 *
    merged_score[
        "deviation_score"
    ]
)

merged_score = (
    merged_score
    .sort_values(
        "Causal_Echo_Score",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\n")
print("="*90)
print("TOP CAUSAL-ECHO CANDIDATES")
print("="*90)

display(
    merged_score[
        [
            "feature",
            "Causal_Echo_Score",
            "first_change_time",
            "maximum_deviation",
            "betweenness",
            "mean_abs_SHAP"
        ]
    ].head(20)
)

# ================================================================
# 31. TOP CANDIDATE VISUALIZATION
# ================================================================

top_candidates = (
    merged_score
    .head(12)
    .sort_values(
        "Causal_Echo_Score"
    )
)

plt.figure(
    figsize=(11,7)
)

plt.barh(

    top_candidates["feature"],

    top_candidates[
        "Causal_Echo_Score"
    ]
)

plt.title(
    "Top Causal-Echo Candidates"
)

plt.xlabel(
    "Causal-Echo Score"
)

plt.tight_layout()
plt.show()

# ================================================================
# 32. ROBUSTNESS TEST — FEATURE NOISE
# ================================================================

noise_levels = [
    0.00,
    0.05,
    0.10,
    0.20,
    0.30
]

robustness_records = []

for noise in noise_levels:

    X_noisy = (
        X_test
        .copy()
    )

    noise_array = np.random.normal(
        0,
        noise,
        X_noisy.shape
    )

    X_noisy = (
        X_noisy +
        noise_array
    )

    noisy_prediction = (
        xgb.predict(
            X_noisy
        )
    )

    noisy_r2 = r2_score(
        y_test,
        noisy_prediction
    )

    noisy_mae = mean_absolute_error(
        y_test,
        noisy_prediction
    )

    robustness_records.append({

        "noise_level":
            noise,

        "R2":
            noisy_r2,

        "MAE":
            noisy_mae
    })

robustness_df = pd.DataFrame(
    robustness_records
)

print("\nNoise robustness")

display(
    robustness_df
)

# ================================================================
# 33. ROBUSTNESS VISUALIZATION
# ================================================================

plt.figure(
    figsize=(10,6)
)

plt.plot(

    robustness_df[
        "noise_level"
    ],

    robustness_df["R2"],

    marker="o",

    linewidth=3
)

plt.title(
    "Model Robustness Under Increasing Feature Noise"
)

plt.xlabel(
    "Noise level"
)

plt.ylabel(
    "R²"
)

plt.grid(
    alpha=0.3
)

plt.show()

# ================================================================
# 34. FEATURE ABLATION EXPERIMENT
# ================================================================
#
# Remove the top SHAP feature and retrain.
# This tests whether the model depends strongly on that feature.
#
# ================================================================

top_feature = (
    shap_df.iloc[0]
    ["feature"]
)

print(
    "\nTop SHAP feature:",
    top_feature
)

X_ablation = (
    X.drop(
        columns=[
            top_feature
        ]
    )
)

X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(

    X_ablation,

    y,

    test_size=0.20,

    random_state=42
)

xgb_ablation = XGBRegressor(

    n_estimators=250,

    max_depth=5,

    learning_rate=0.04,

    subsample=0.85,

    colsample_bytree=0.85,

    random_state=42,

    n_jobs=-1
)

xgb_ablation.fit(
    X_train_a,
    y_train_a
)

ablation_pred = (
    xgb_ablation.predict(
        X_test_a
    )
)

ablation_r2 = r2_score(
    y_test_a,
    ablation_pred
)

print(
    "Original XGBoost R²:",
    round(xgb_r2,4)
)

print(
    "After removing top feature:",
    round(ablation_r2,4)
)

print(
    "R² change:",
    round(
        ablation_r2 - xgb_r2,
        4
    )
)

# ================================================================
# 35. ROOT-CAUSE CANDIDATE ANALYSIS
# ================================================================

# Features that:
# 1. Change early
# 2. Have strong downstream network influence
# 3. Are important to ML
# 4. Show strong deviation

root_candidates = (
    merged_score
    .sort_values(
        "Causal_Echo_Score",
        ascending=False
    )
    .head(10)
)

print("\n")
print("="*90)
print("POTENTIAL EARLY-DISTURBANCE CANDIDATES")
print("="*90)

for i, (_, row) in enumerate(
    root_candidates.iterrows(),
    start=1
):

    print(
        f"\n{i}. {row['feature']}"
    )

    print(
        "Causal-Echo Score:",
        round(
            row[
                "Causal_Echo_Score"
            ],
            4
        )
    )

    print(
        "First change:",
        row[
            "first_change_time"
        ]
    )

    print(
        "Network betweenness:",
        round(
            row[
                "betweenness"
            ],
            4
        )
    )

    print(
        "SHAP importance:",
        round(
            row[
                "mean_abs_SHAP"
            ],
            4
        )
    )

# ================================================================
# 36. AUTOMATIC RESEARCH HYPOTHESES
# ================================================================

print("\n")
print("="*90)
print("AUTOMATICALLY GENERATED COMPUTATIONAL HYPOTHESES")
print("="*90)

for i, (_, row) in enumerate(
    root_candidates.head(8).iterrows(),
    start=1
):

    feature = row["feature"]

    print(
        f"\nHypothesis {i}"
    )

    print(
        f"{feature} appears as an early-changing "
        f"and computationally important variable."
    )

    print(
        "Potential interpretation:"
    )

    print(
        f"The temporal, network, and ML signals for "
        f"{feature} make it a candidate for further "
        f"investigation as part of the simulated "
        f"propagation cascade."
    )

    print(
        "Required validation:"
    )

    print(
        "Real longitudinal data + intervention/perturbation "
        "experiments + independent biological evidence."
    )

# ================================================================
# 37. SAVE RESEARCH OUTPUTS
# ================================================================

data.to_csv(
    "CausalEcho_temporal_biological_data.csv",
    index=False
)

summary.to_csv(
    "CausalEcho_feature_statistics.csv"
)

temporal_df.to_csv(
    "CausalEcho_temporal_precedence.csv",
    index=False
)

model_results.to_csv(
    "CausalEcho_model_comparison.csv",
    index=False
)

importance_df.to_csv(
    "CausalEcho_XGBoost_importance.csv",
    index=False
)

permutation_df.to_csv(
    "CausalEcho_permutation_importance.csv",
    index=False
)

shap_df.to_csv(
    "CausalEcho_SHAP_importance.csv",
    index=False
)

lag_df.to_csv(
    "CausalEcho_lagged_relationships.csv",
    index=False
)

edge_evidence_df.to_csv(
    "CausalEcho_discovered_edges.csv",
    index=False
)

network_df.to_csv(
    "CausalEcho_network_centrality.csv",
    index=False
)

merged_score.to_csv(
    "CausalEcho_Causal_Echo_Scores.csv",
    index=False
)

transitions_df.to_csv(
    "CausalEcho_state_transitions.csv",
    index=False
)

robustness_df.to_csv(
    "CausalEcho_robustness_analysis.csv",
    index=False
)

root_candidates.to_csv(
    "CausalEcho_root_candidates.csv",
    index=False
)

# ================================================================
# 38. FINAL RESEARCH DASHBOARD
# ================================================================

print("\n")
print("="*90)
print("CAUSAL-ECHO AI — FINAL RESEARCH DASHBOARD")
print("="*90)

print(
    "\nDataset:"
)

print(
    "Samples:",
    N_SAMPLES
)

print(
    "Temporal observations:",
    len(data)
)

print(
    "Biological variables:",
    N_BIO_FEATURES
)

print(
    "\nMachine Learning:"
)

print(
    "Random Forest R²:",
    round(rf_r2,4)
)

print(
    "XGBoost R²:",
    round(xgb_r2,4)
)

print(
    "\nCausal-structure recovery:"
)

print(
    "True edges:",
    len(true_edge_set)
)

print(
    "Discovered edges:",
    len(discovered_edges)
)

print(
    "Recovered true edges:",
    len(true_edges_found)
)

print(
    "Structural precision:",
    round(
        precision_discovery,
        4
    )
)

print(
    "Structural recall:",
    round(
        recall_discovery,
        4
    )
)

print(
    "\nTop Causal-Echo candidate:"
)

print(
    merged_score.iloc[0]["feature"]
)

print(
    "Score:",
    round(
        merged_score.iloc[0][
            "Causal_Echo_Score"
        ],
        4
    )
)

print(
    "\nTop SHAP feature:"
)

print(
    shap_df.iloc[0]["feature"]
)

print(
    "\nTop temporal early-change feature:"
)

print(
    temporal_df.iloc[0]["feature"]
)

print(
    "\nFiles generated:"
)

output_files = [

    "CausalEcho_temporal_biological_data.csv",
    "CausalEcho_feature_statistics.csv",
    "CausalEcho_temporal_precedence.csv",
    "CausalEcho_model_comparison.csv",
    "CausalEcho_XGBoost_importance.csv",
    "CausalEcho_permutation_importance.csv",
    "CausalEcho_SHAP_importance.csv",
    "CausalEcho_lagged_relationships.csv",
    "CausalEcho_discovered_edges.csv",
    "CausalEcho_network_centrality.csv",
    "CausalEcho_Causal_Echo_Scores.csv",
    "CausalEcho_state_transitions.csv",
    "CausalEcho_robustness_analysis.csv",
    "CausalEcho_root_candidates.csv"
]

for filename in output_files:

    print(
        "✓",
        filename
    )

# ================================================================
# 39. SCIENTIFIC LIMITATIONS
# ================================================================

print("\n")
print("="*90)
print("IMPORTANT SCIENTIFIC LIMITATIONS")
print("="*90)

print("""
1. This notebook uses SIMULATED biological-style data.

2. The hidden causal graph is known because it was used to
   generate the synthetic dataset.

3. The discovered relationships therefore demonstrate an
   algorithmic experiment, not a biological discovery.

4. Temporal precedence and lagged correlation do NOT establish
   biological causality by themselves.

5. The Causal-Echo Score is a computational prioritization
   heuristic, not a formal causal estimator.

6. SHAP measures model attribution, not biological causation.

7. Network centrality describes graph structure and does not
   automatically imply biological importance.

8. Real biological applications require carefully designed
   observational or intervention datasets.

9. Real-world validation should include:
   - longitudinal experiments
   - perturbation/intervention data
   - biological replicates
   - batch-effect correction
   - independent datasets
   - pathway annotation
   - literature evidence
   - experimental validation

10. The next research stage would be to replace the simulated
    system with real transcriptomic, proteomic, metabolomic,
    single-cell, or perturbation datasets.
""")

# ================================================================
# 40. COMPLETION
# ================================================================

print("\n")
print("="*90)
print("CAUSAL-ECHO AI COMPLETED SUCCESSFULLY")
print("="*90)

print("""
Research pipeline completed:

Synthetic Biology
        ↓
Temporal Perturbation
        ↓
Statistical Analysis
        ↓
Random Forest
        ↓
XGBoost
        ↓
Feature Importance
        ↓
SHAP
        ↓
Anomaly Detection
        ↓
Hidden State Discovery
        ↓
Temporal Precedence
        ↓
Lagged Relationships
        ↓
Propagation Network
        ↓
Network Centrality
        ↓
Structure Recovery
        ↓
Robustness Testing
        ↓
Feature Ablation
        ↓
Causal-Echo Score
        ↓
Root-Cause Candidates
        ↓
Computational Hypotheses

CAUSAL-ECHO AI: COMPLETE
""")# ================================================================
# CAUSAL-ECHO AI
# Reconstructing Hidden Propagation Pathways in Complex Systems
#
# COMPLETE ONE-CELL GOOGLE COLAB RESEARCH PROTOTYPE
# ================================================================

!pip -q install numpy pandas matplotlib seaborn scikit-learn networkx scipy xgboost shap

# ================================================================
# 1. IMPORTS
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx

from itertools import combinations

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.ensemble import (
    RandomForestRegressor,
    RandomForestClassifier,
    IsolationForest
)
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    silhouette_score
)
from sklearn.inspection import permutation_importance

from scipy.stats import pearsonr, spearmanr

from xgboost import XGBRegressor, XGBClassifier

import shap
import warnings

warnings.filterwarnings("ignore")

np.random.seed(42)

print("="*90)
print("CAUSAL-ECHO AI")
print("AI-Based Reconstruction of Hidden Propagation Pathways")
print("="*90)

# ================================================================
# 2. PROJECT CONFIGURATION
# ================================================================

N_SAMPLES = 2500
N_TIMEPOINTS = 10
N_BIO_FEATURES = 25
NOISE_LEVEL = 0.08

feature_names = [
    "Gene_Expression_A",
    "Gene_Expression_B",
    "Gene_Expression_C",
    "Gene_Expression_D",
    "Gene_Expression_E",
    "Protein_Activity_A",
    "Protein_Activity_B",
    "Protein_Activity_C",
    "Metabolic_Activity",
    "ATP_Level",
    "ROS_Level",
    "Oxidative_Stress",
    "Inflammation",
    "DNA_Damage",
    "DNA_Repair",
    "Mitochondrial_Activity",
    "Protein_Synthesis",
    "Autophagy",
    "Cell_Cycle",
    "Apoptosis_Signal",
    "Immune_Signaling",
    "Growth_Signaling",
    "Membrane_Integrity",
    "Cellular_Stress",
    "Final_Cell_State"
]

print("\nConfiguration")
print("-"*60)
print("Samples:", N_SAMPLES)
print("Timepoints:", N_TIMEPOINTS)
print("Biological variables:", N_BIO_FEATURES)
print("Noise level:", NOISE_LEVEL)

# ================================================================
# 3. HIDDEN BIOLOGICAL CAUSAL STRUCTURE
# ================================================================
#
# IMPORTANT:
# This structure is hidden from the ML discovery algorithms.
# It is used only to generate synthetic ground truth.
#
# Example:
#
# Initial_Perturbation
#        ↓
# Gene A
#        ↓
# Gene B
#        ↓
# Protein A
#        ↓
# ROS
#        ↓
# Oxidative Stress
#        ↓
# Mitochondrial Activity
#        ↓
# ATP
#        ↓
# Cellular Stress
#        ↓
# Final Cell State
#
# ================================================================

true_edges = [

    ("Gene_Expression_A", "Gene_Expression_B"),
    ("Gene_Expression_B", "Gene_Expression_C"),
    ("Gene_Expression_C", "Protein_Activity_A"),

    ("Protein_Activity_A", "ROS_Level"),
    ("ROS_Level", "Oxidative_Stress"),

    ("Oxidative_Stress", "Mitochondrial_Activity"),
    ("Mitochondrial_Activity", "ATP_Level"),

    ("ATP_Level", "Cellular_Stress"),

    ("Oxidative_Stress", "DNA_Damage"),
    ("DNA_Damage", "DNA_Repair"),

    ("DNA_Damage", "Apoptosis_Signal"),

    ("Gene_Expression_D", "Inflammation"),
    ("Inflammation", "Immune_Signaling"),

    ("Gene_Expression_E", "Growth_Signaling"),

    ("Protein_Activity_B", "Protein_Synthesis"),
    ("Protein_Synthesis", "Protein_Quality_Control")
    if "Protein_Quality_Control" in feature_names
    else ("Protein_Synthesis", "Autophagy"),

    ("ATP_Level", "Cell_Cycle"),
    ("Cell_Cycle", "Apoptosis_Signal"),

    ("Autophagy", "Cellular_Stress"),
    ("Immune_Signaling", "Cellular_Stress"),

    ("Cellular_Stress", "Membrane_Integrity"),
    ("Membrane_Integrity", "Final_Cell_State"),

    ("Apoptosis_Signal", "Final_Cell_State"),
    ("Growth_Signaling", "Final_Cell_State"),
    ("ATP_Level", "Final_Cell_State")
]

# Remove invalid self/nonexistent edges
true_edges = [
    (a,b)
    for a,b in true_edges
    if a in feature_names and b in feature_names and a != b
]

true_edge_set = set(true_edges)

print("\nHidden causal structure contains:")
print(len(true_edges), "ground-truth edges")

# ================================================================
# 4. CREATE DIRECTED GROUND-TRUTH GRAPH
# ================================================================

true_graph = nx.DiGraph()

true_graph.add_nodes_from(
    feature_names
)

true_graph.add_edges_from(
    true_edges
)

print(
    "Ground-truth nodes:",
    true_graph.number_of_nodes()
)

print(
    "Ground-truth edges:",
    true_graph.number_of_edges()
)

# ================================================================
# 5. GENERATE TEMPORAL BIOLOGICAL DATA
# ================================================================
#
# Each sample has a perturbation intensity.
# The hidden causal chain propagates across time.
#
# ================================================================

records = []

for sample_id in range(N_SAMPLES):

    # ------------------------------------------------------------
    # Initial perturbation
    # ------------------------------------------------------------

    perturbation = np.random.normal(
        0,
        1
    )

    baseline = np.random.normal(
        0,
        0.15,
        N_BIO_FEATURES
    )

    values = {}

    # ------------------------------------------------------------
    # Time-dependent perturbation
    # ------------------------------------------------------------

    for t in range(N_TIMEPOINTS):

        progression = (
            perturbation *
            (t / (N_TIMEPOINTS - 1))
        )

        # --------------------------------------------------------
        # Root biological variables
        # --------------------------------------------------------

        A = (
            baseline[0]
            + progression * 1.4
            + np.random.normal(0, NOISE_LEVEL)
        )

        B = (
            baseline[1]
            + 0.70*A
            + progression*0.15
            + np.random.normal(0, NOISE_LEVEL)
        )

        C = (
            baseline[2]
            + 0.75*B
            + np.random.normal(0, NOISE_LEVEL)
        )

        D = (
            baseline[3]
            + np.random.normal(0, NOISE_LEVEL)
        )

        E = (
            baseline[4]
            + np.random.normal(0, NOISE_LEVEL)
        )

        Protein_A = (
            baseline[5]
            + 0.85*C
            + np.random.normal(0, NOISE_LEVEL)
        )

        Protein_B = (
            baseline[6]
            + np.random.normal(0, NOISE_LEVEL)
        )

        Protein_C = (
            baseline[7]
            + np.random.normal(0, NOISE_LEVEL)
        )

        # --------------------------------------------------------
        # Metabolic pathway
        # --------------------------------------------------------

        ROS = (
            baseline[10]
            + 0.80*Protein_A
            + 0.25*progression
            + np.random.normal(0, NOISE_LEVEL)
        )

        OxStress = (
            baseline[11]
            + 0.85*ROS
            + np.random.normal(0, NOISE_LEVEL)
        )

        Mito = (
            baseline[15]
            - 0.70*OxStress
            + np.random.normal(0, NOISE_LEVEL)
        )

        ATP = (
            baseline[9]
            + 0.80*Mito
            - 0.25*OxStress
            + np.random.normal(0, NOISE_LEVEL)
        )

        # --------------------------------------------------------
        # DNA pathway
        # --------------------------------------------------------

        DNA_damage = (
            baseline[13]
            + 0.75*OxStress
            + np.random.normal(0, NOISE_LEVEL)
        )

        DNA_repair = (
            baseline[14]
            + 0.65*DNA_damage
            + np.random.normal(0, NOISE_LEVEL)
        )

        # --------------------------------------------------------
        # Immune pathway
        # --------------------------------------------------------

        Inflammation = (
            baseline[12]
            + 0.70*D
            + 0.20*progression
            + np.random.normal(0, NOISE_LEVEL)
        )

        Immune = (
            baseline[20]
            + 0.75*Inflammation
            + np.random.normal(0, NOISE_LEVEL)
        )

        # --------------------------------------------------------
        # Growth pathway
        # --------------------------------------------------------

        Growth = (
            baseline[21]
            + 0.80*E
            + np.random.normal(0, NOISE_LEVEL)
        )

        # --------------------------------------------------------
        # Protein / cellular maintenance
        # --------------------------------------------------------

        Protein_Synthesis = (
            baseline[16]
            + 0.60*Protein_B
            + 0.25*ATP
            + np.random.normal(0, NOISE_LEVEL)
        )

        Autophagy = (
            baseline[17]
            + 0.55*OxStress
            - 0.25*ATP
            + np.random.normal(0, NOISE_LEVEL)
        )

        Cell_Cycle = (
            baseline[18]
            + 0.55*ATP
            + 0.30*Growth
            + np.random.normal(0, NOISE_LEVEL)
        )

        Apoptosis = (
            baseline[19]
            + 0.70*DNA_damage
            + 0.50*OxStress
            - 0.30*ATP
            + np.random.normal(0, NOISE_LEVEL)
        )

        Membrane = (
            baseline[22]
            + 0.55*ATP
            - 0.60*OxStress
            - 0.50*Apoptosis
            + np.random.normal(0, NOISE_LEVEL)
        )

        CellularStress = (
            baseline[23]
            + 0.55*OxStress
            + 0.50*Inflammation
            + 0.40*Autophagy
            - 0.40*ATP
            + np.random.normal(0, NOISE_LEVEL)
        )

        FinalState = (
            0.45*CellularStress
            + 0.35*Apoptosis
            + 0.30*Growth
            + 0.25*ATP
            + np.random.normal(0, NOISE_LEVEL)
        )

        row = {

            "sample_id":
                sample_id,

            "time":
                t,

            "perturbation":
                perturbation,

            "Gene_Expression_A":
                A,

            "Gene_Expression_B":
                B,

            "Gene_Expression_C":
                C,

            "Gene_Expression_D":
                D,

            "Gene_Expression_E":
                E,

            "Protein_Activity_A":
                Protein_A,

            "Protein_Activity_B":
                Protein_B,

            "Protein_Activity_C":
                Protein_C,

            "Metabolic_Activity":
                0.6*ATP + np.random.normal(0, NOISE_LEVEL),

            "ATP_Level":
                ATP,

            "ROS_Level":
                ROS,

            "Oxidative_Stress":
                OxStress,

            "Inflammation":
                Inflammation,

            "DNA_Damage":
                DNA_damage,

            "DNA_Repair":
                DNA_repair,

            "Mitochondrial_Activity":
                Mito,

            "Protein_Synthesis":
                Protein_Synthesis,

            "Autophagy":
                Autophagy,

            "Cell_Cycle":
                Cell_Cycle,

            "Apoptosis_Signal":
                Apoptosis,

            "Immune_Signaling":
                Immune,

            "Growth_Signaling":
                Growth,

            "Membrane_Integrity":
                Membrane,

            "Cellular_Stress":
                CellularStress,

            "Final_Cell_State":
                FinalState
        }

        records.append(row)

data = pd.DataFrame(
    records
)

print("\nGenerated temporal biological dataset:")
print(data.shape)

display(
    data.head()
)

# ================================================================
# 6. DATA QUALITY ANALYSIS
# ================================================================

print("\nDATA QUALITY")
print("-"*60)

print(
    "Missing values:",
    data.isna().sum().sum()
)

print(
    "Duplicate rows:",
    data.duplicated().sum()
)

print(
    "Samples:",
    data["sample_id"].nunique()
)

print(
    "Timepoints:",
    data["time"].nunique()
)

# ================================================================
# 7. BASIC STATISTICAL SUMMARY
# ================================================================

summary = data[
    feature_names
].describe().T

summary[
    "variance"
] = data[
    feature_names
].var()

display(
    summary.head(15)
)

# ================================================================
# 8. CORRELATION NETWORK
# ================================================================

corr = data[
    feature_names
].corr(
    method="spearman"
)

plt.figure(
    figsize=(18,14)
)

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0,
    square=True
)

plt.title(
    "Biological Feature Correlation Matrix"
)

plt.tight_layout()
plt.show()

# ================================================================
# 9. TEMPORAL FEATURE TRAJECTORIES
# ================================================================

important_features = [

    "Gene_Expression_A",
    "Gene_Expression_B",
    "Protein_Activity_A",
    "ROS_Level",
    "Oxidative_Stress",
    "ATP_Level",
    "DNA_Damage",
    "Cellular_Stress",
    "Final_Cell_State"
]

plt.figure(
    figsize=(15,8)
)

for feature in important_features:

    temporal_mean = (
        data
        .groupby("time")[feature]
        .mean()
    )

    plt.plot(
        temporal_mean.index,
        temporal_mean.values,
        marker="o",
        linewidth=2,
        label=feature
    )

plt.title(
    "Temporal Biological Cascade"
)

plt.xlabel(
    "Time"
)

plt.ylabel(
    "Mean standardized biological signal"
)

plt.legend(
    bbox_to_anchor=(1.02,1),
    loc="upper left"
)

plt.grid(
    alpha=0.25
)

plt.tight_layout()
plt.show()

# ================================================================
# 10. TEMPORAL PRECEDENCE ANALYSIS
# ================================================================
#
# A simple computational heuristic:
# measure when each feature starts deviating strongly
# from its baseline.
#
# This is NOT proof of causality.
#
# ================================================================

temporal_records = []

for feature in feature_names:

    means = (
        data
        .groupby("time")[feature]
        .mean()
        .reindex(
            range(N_TIMEPOINTS)
        )
    )

    baseline = means.iloc[0]

    deviations = (
        abs(means - baseline)
    )

    max_deviation = deviations.max()

    threshold = (
        0.30 *
        max_deviation
    )

    candidate_times = (
        deviations[
            deviations >= threshold
        ]
        .index
    )

    first_change = (
        candidate_times[0]
        if len(candidate_times) > 0
        else N_TIMEPOINTS
    )

    temporal_records.append({

        "feature":
            feature,

        "first_change_time":
            first_change,

        "maximum_deviation":
            max_deviation,

        "final_change":
            means.iloc[-1] - baseline
    })

temporal_df = pd.DataFrame(
    temporal_records
)

temporal_df = (
    temporal_df
    .sort_values(
        "first_change_time"
    )
    .reset_index(drop=True)
)

print("\nTemporal precedence analysis")

display(
    temporal_df
)

# ================================================================
# 11. MACHINE LEARNING — RANDOM FOREST REGRESSION
# ================================================================
#
# Predict Final_Cell_State using all other biological features.
#
# ================================================================

target = "Final_Cell_State"

predictor_features = [
    f for f in feature_names
    if f != target
]

X = data[
    predictor_features
]

y = data[
    target
]

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42
)

rf = RandomForestRegressor(

    n_estimators=250,

    max_depth=12,

    min_samples_leaf=3,

    random_state=42,

    n_jobs=-1
)

rf.fit(
    X_train,
    y_train
)

rf_pred = rf.predict(
    X_test
)

rf_mae = mean_absolute_error(
    y_test,
    rf_pred
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_pred
    )
)

rf_r2 = r2_score(
    y_test,
    rf_pred
)

print("\nRandom Forest")

print(
    "MAE:",
    round(rf_mae,4)
)

print(
    "RMSE:",
    round(rf_rmse,4)
)

print(
    "R²:",
    round(rf_r2,4)
)

# ================================================================
# 12. XGBOOST MODEL
# ================================================================

xgb = XGBRegressor(

    n_estimators=300,

    max_depth=5,

    learning_rate=0.04,

    subsample=0.85,

    colsample_bytree=0.85,

    objective="reg:squarederror",

    random_state=42,

    n_jobs=-1
)

xgb.fit(
    X_train,
    y_train
)

xgb_pred = xgb.predict(
    X_test
)

xgb_mae = mean_absolute_error(
    y_test,
    xgb_pred
)

xgb_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        xgb_pred
    )
)

xgb_r2 = r2_score(
    y_test,
    xgb_pred
)

print("\nXGBoost")

print(
    "MAE:",
    round(xgb_mae,4)
)

print(
    "RMSE:",
    round(xgb_rmse,4)
)

print(
    "R²:",
    round(xgb_r2,4)
)

# ================================================================
# 13. MODEL COMPARISON
# ================================================================

model_results = pd.DataFrame({

    "model": [
        "Random Forest",
        "XGBoost"
    ],

    "MAE": [
        rf_mae,
        xgb_mae
    ],

    "RMSE": [
        rf_rmse,
        xgb_rmse
    ],

    "R2": [
        rf_r2,
        xgb_r2
    ]
})

display(
    model_results
)

# ================================================================
# 14. ACTUAL VS PREDICTED
# ================================================================

plt.figure(
    figsize=(9,7)
)

plt.scatter(
    y_test,
    xgb_pred,
    alpha=0.5
)

minimum = min(
    y_test.min(),
    xgb_pred.min()
)

maximum = max(
    y_test.max(),
    xgb_pred.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.xlabel(
    "Actual Final Cell State"
)

plt.ylabel(
    "Predicted Final Cell State"
)

plt.title(
    "XGBoost — Actual vs Predicted"
)

plt.grid(
    alpha=0.25
)

plt.show()

# ================================================================
# 15. FEATURE IMPORTANCE
# ================================================================

importance_df = pd.DataFrame({

    "feature":
        predictor_features,

    "importance":
        xgb.feature_importances_
})

importance_df = (
    importance_df
    .sort_values(
        "importance",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nXGBoost Feature Importance")

display(
    importance_df.head(20)
)

plt.figure(
    figsize=(12,8)
)

top_imp = (
    importance_df
    .head(15)
    .sort_values(
        "importance"
    )
)

plt.barh(
    top_imp["feature"],
    top_imp["importance"]
)

plt.title(
    "Top Features Driving Final Cell-State Prediction"
)

plt.xlabel(
    "XGBoost Importance"
)

plt.tight_layout()
plt.show()

# ================================================================
# 16. PERMUTATION IMPORTANCE
# ================================================================

perm = permutation_importance(

    xgb,

    X_test,

    y_test,

    n_repeats=5,

    random_state=42,

    scoring="r2"
)

permutation_df = pd.DataFrame({

    "feature":
        predictor_features,

    "permutation_importance":
        perm.importances_mean
})

permutation_df = (
    permutation_df
    .sort_values(
        "permutation_importance",
        ascending=False
    )
)

print("\nPermutation importance")

display(
    permutation_df.head(15)
)

# ================================================================
# 17. SHAP EXPLAINABILITY
# ================================================================

print("\nCalculating SHAP explanations...")

explainer = shap.TreeExplainer(
    xgb
)

sample_for_shap = (
    X_test
    .sample(
        min(500, len(X_test)),
        random_state=42
    )
)

shap_values = explainer(
    sample_for_shap
)

plt.figure(
    figsize=(12,8)
)

shap.summary_plot(
    shap_values.values,
    sample_for_shap,
    show=False
)

plt.title(
    "SHAP — Global Model Explanation"
)

plt.tight_layout()
plt.show()

# ================================================================
# 18. SHAP IMPORTANCE TABLE
# ================================================================

mean_abs_shap = np.abs(
    shap_values.values
).mean(
    axis=0
)

shap_df = pd.DataFrame({

    "feature":
        predictor_features,

    "mean_abs_SHAP":
        mean_abs_shap
})

shap_df = (
    shap_df
    .sort_values(
        "mean_abs_SHAP",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nSHAP Feature Importance")

display(
    shap_df.head(20)
)

# ================================================================
# 19. ANOMALY DETECTION
# ================================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    data[
        predictor_features
    ]
)

isolation = IsolationForest(

    n_estimators=250,

    contamination=0.05,

    random_state=42
)

anomaly_labels = isolation.fit_predict(
    X_scaled
)

anomaly_scores = -isolation.score_samples(
    X_scaled
)

data["anomaly_label"] = anomaly_labels

data["anomaly_score"] = anomaly_scores

print("\nAnomaly detection")

print(
    "Detected anomalies:",
    (anomaly_labels == -1).sum()
)

# ================================================================
# 20. ANOMALY SCORE THROUGH TIME
# ================================================================

anomaly_temporal = (
    data
    .groupby("time")["anomaly_score"]
    .mean()
)

plt.figure(
    figsize=(11,6)
)

plt.plot(
    anomaly_temporal.index,
    anomaly_temporal.values,
    marker="o",
    linewidth=3
)

plt.title(
    "System Anomaly Intensity Through Time"
)

plt.xlabel(
    "Time"
)

plt.ylabel(
    "Mean anomaly score"
)

plt.grid(
    alpha=0.3
)

plt.show()

# ================================================================
# 21. HIDDEN STATE DISCOVERY
# ================================================================

state_model = KMeans(

    n_clusters=4,

    random_state=42,

    n_init=20
)

state_labels = state_model.fit_predict(
    X_scaled
)

data["hidden_state"] = state_labels

state_profile = (
    data
    .groupby("hidden_state")[
        predictor_features
    ]
    .mean()
)

print("\nHidden biological states")

display(
    state_profile
)

# ================================================================
# 22. PCA STATE VISUALIZATION
# ================================================================

pca = PCA(
    n_components=2,
    random_state=42
)

pca_data = pca.fit_transform(
    X_scaled
)

plt.figure(
    figsize=(11,8)
)

plt.scatter(

    pca_data[:,0],

    pca_data[:,1],

    c=data["hidden_state"],

    cmap="tab10",

    alpha=0.55,

    s=20
)

plt.xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)"
)

plt.ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)"
)

plt.title(
    "Hidden Biological State Landscape"
)

plt.grid(
    alpha=0.25
)

plt.show()

# ================================================================
# 23. TEMPORAL STATE TRANSITION MATRIX
# ================================================================

state_transition_records = []

for sample_id in data[
    "sample_id"
].unique():

    sample = (
        data[
            data["sample_id"] == sample_id
        ]
        .sort_values("time")
    )

    states = sample[
        "hidden_state"
    ].values

    for i in range(
        len(states)-1
    ):

        state_transition_records.append({

            "sample_id":
                sample_id,

            "time":
                i,

            "from_state":
                states[i],

            "to_state":
                states[i+1]
        })

transitions_df = pd.DataFrame(
    state_transition_records
)

transition_matrix = pd.crosstab(

    transitions_df[
        "from_state"
    ],

    transitions_df[
        "to_state"
    ]
)

print("\nHidden-state transition matrix")

display(
    transition_matrix
)

plt.figure(
    figsize=(8,7)
)

sns.heatmap(
    transition_matrix,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Biological State Transition Matrix"
)

plt.xlabel(
    "To State"
)

plt.ylabel(
    "From State"
)

plt.show()

# ================================================================
# 24. LAGGED RELATIONSHIP ANALYSIS
# ================================================================
#
# Check whether changes in feature A at time t are associated
# with feature B at time t+1.
#
# This provides temporal evidence but does NOT prove causality.
#
# ================================================================

lag_records = []

for source, target_feature in combinations(
    feature_names,
    2
):

    source_values = []
    target_values = []

    for sample_id in data[
        "sample_id"
    ].unique():

        sample = (
            data[
                data["sample_id"] == sample_id
            ]
            .sort_values("time")
        )

        if len(sample) < 2:
            continue

        source_values.extend(
            sample[source]
            .iloc[:-1]
            .values
        )

        target_values.extend(
            sample[target_feature]
            .iloc[1:]
            .values
        )

    if len(source_values) > 10:

        correlation, pvalue = pearsonr(
            source_values,
            target_values
        )

        lag_records.append({

            "source":
                source,

            "target":
                target_feature,

            "lag_correlation":
                correlation,

            "p_value":
                pvalue
        })

lag_df = pd.DataFrame(
    lag_records
)

lag_df["abs_lag_correlation"] = (
    abs(
        lag_df[
            "lag_correlation"
        ]
    )
)

lag_df = (
    lag_df
    .sort_values(
        "abs_lag_correlation",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nStrongest lagged relationships")

display(
    lag_df.head(25)
)

# ================================================================
# 25. DISCOVERED PROPAGATION GRAPH
# ================================================================

DISCOVERY_THRESHOLD = 0.35

discovered_graph = nx.DiGraph()

discovered_graph.add_nodes_from(
    feature_names
)

for _, row in lag_df.iterrows():

    if (
        abs(
            row["lag_correlation"]
        )
        >= DISCOVERY_THRESHOLD
    ):

        discovered_graph.add_edge(

            row["source"],

            row["target"],

            weight=abs(
                row["lag_correlation"]
            ),

            signed_weight=
                row["lag_correlation"]
        )

print("\nDiscovered propagation network")

print(
    "Nodes:",
    discovered_graph.number_of_nodes()
)

print(
    "Edges:",
    discovered_graph.number_of_edges()
)

# ================================================================
# 26. COMPARE DISCOVERED EDGES WITH HIDDEN TRUTH
# ================================================================

discovered_edges = set(
    discovered_graph.edges()
)

true_edges_found = (
    discovered_edges
    .intersection(
        true_edge_set
    )
)

false_edges = (
    discovered_edges
    - true_edge_set
)

missed_edges = (
    true_edge_set
    - discovered_edges
)

precision_discovery = (
    len(true_edges_found) /
    max(len(discovered_edges),1)
)

recall_discovery = (
    len(true_edges_found) /
    max(len(true_edge_set),1)
)

print("\nHidden-structure recovery")

print(
    "True edges:",
    len(true_edge_set)
)

print(
    "Discovered edges:",
    len(discovered_edges)
)

print(
    "Correctly recovered:",
    len(true_edges_found)
)

print(
    "False edges:",
    len(false_edges)
)

print(
    "Missed true edges:",
    len(missed_edges)
)

print(
    "Structural precision:",
    round(
        precision_discovery,
        4
    )
)

print(
    "Structural recall:",
    round(
        recall_discovery,
        4
    )
)

# ================================================================
# 27. DISCOVERED NETWORK VISUALIZATION
# ================================================================

plt.figure(
    figsize=(16,12)
)

pos = nx.spring_layout(
    discovered_graph,
    seed=42,
    k=0.55
)

nx.draw_networkx_nodes(

    discovered_graph,

    pos,

    node_size=900,

    alpha=0.85
)

nx.draw_networkx_edges(

    discovered_graph,

    pos,

    arrows=True,

    arrowsize=15,

    alpha=0.25,

    width=1.5,

    connectionstyle="arc3,rad=0.05"
)

nx.draw_networkx_labels(

    discovered_graph,

    pos,

    font_size=8
)

plt.title(
    "CAUSAL-ECHO — Discovered Temporal Propagation Network",
    fontsize=17
)

plt.axis("off")
plt.show()

# ================================================================
# 28. EDGE-LEVEL EVIDENCE TABLE
# ================================================================

edge_records = []

for source, target in discovered_graph.edges():

    lag_value = lag_df[
        (
            lag_df["source"] == source
        )
        &
        (
            lag_df["target"] == target
        )
    ]

    if len(lag_value) == 0:
        continue

    correlation_value = (
        lag_value.iloc[0]
        ["lag_correlation"]
    )

    is_true = (
        (source, target)
        in true_edge_set
    )

    edge_records.append({

        "source":
            source,

        "target":
            target,

        "lag_correlation":
            correlation_value,

        "absolute_strength":
            abs(correlation_value),

        "ground_truth_edge":
            is_true
    })

edge_evidence_df = pd.DataFrame(
    edge_records
)

edge_evidence_df = (
    edge_evidence_df
    .sort_values(
        "absolute_strength",
        ascending=False
    )
)

display(
    edge_evidence_df.head(30)
)

# ================================================================
# 29. NETWORK CENTRALITY
# ================================================================

if discovered_graph.number_of_edges() > 0:

    in_degree = dict(
        discovered_graph.in_degree()
    )

    out_degree = dict(
        discovered_graph.out_degree()
    )

    centrality = nx.betweenness_centrality(
        discovered_graph
    )

else:

    in_degree = {
        node:0
        for node in feature_names
    }

    out_degree = {
        node:0
        for node in feature_names
    }

    centrality = {
        node:0
        for node in feature_names
    }

network_df = pd.DataFrame({

    "feature":
        feature_names,

    "in_degree":
        [
            in_degree.get(
                f,
                0
            )
            for f in feature_names
        ],

    "out_degree":
        [
            out_degree.get(
                f,
                0
            )
            for f in feature_names
        ],

    "betweenness":
        [
            centrality.get(
                f,
                0
            )
            for f in feature_names
        ]
})

print("\nNetwork-central features")

display(
    network_df
    .sort_values(
        "betweenness",
        ascending=False
    )
    .head(15)
)

# ================================================================
# 30. TEMPORAL + NETWORK + ML SCORE
# ================================================================
#
# Causal-Echo Score is a research prioritization heuristic.
# It does NOT establish causality.
#
# ================================================================

merged_score = (
    temporal_df[
        [
            "feature",
            "first_change_time",
            "maximum_deviation",
            "final_change"
        ]
    ]
    .merge(
        network_df,
        on="feature",
        how="left"
    )
)

merged_score = (
    merged_score
    .merge(
        shap_df.rename(
            columns={
                "feature":
                    "feature"
            }
        ),
        on="feature",
        how="left"
    )
)

# Earlier change = higher temporal score
merged_score[
    "temporal_score"
] = (

    1 -

    (
        merged_score[
            "first_change_time"
        ]
        /
        max(
            N_TIMEPOINTS,
            1
        )
    )
)

merged_score[
    "temporal_score"
] = (
    merged_score[
        "temporal_score"
    ]
    .clip(0,1)
)

# Normalize network score
def minmax(series):

    minimum = series.min()
    maximum = series.max()

    if maximum == minimum:
        return pd.Series(
            0.5,
            index=series.index
        )

    return (
        (series - minimum)
        /
        (maximum - minimum)
    )

merged_score[
    "network_score"
] = minmax(
    merged_score[
        "betweenness"
    ]
)

merged_score[
    "SHAP_score"
] = minmax(
    merged_score[
        "mean_abs_SHAP"
    ]
)

merged_score[
    "deviation_score"
] = minmax(
    merged_score[
        "maximum_deviation"
    ]
)

# ---------------------------------------------------------------
# Causal-Echo Score
# ---------------------------------------------------------------

merged_score[
    "Causal_Echo_Score"
] = (

    0.30 *
    merged_score[
        "temporal_score"
    ]

    +

    0.25 *
    merged_score[
        "network_score"
    ]

    +

    0.25 *
    merged_score[
        "SHAP_score"
    ]

    +

    0.20 *
    merged_score[
        "deviation_score"
    ]
)

merged_score = (
    merged_score
    .sort_values(
        "Causal_Echo_Score",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\n")
print("="*90)
print("TOP CAUSAL-ECHO CANDIDATES")
print("="*90)

display(
    merged_score[
        [
            "feature",
            "Causal_Echo_Score",
            "first_change_time",
            "maximum_deviation",
            "betweenness",
            "mean_abs_SHAP"
        ]
    ].head(20)
)

# ================================================================
# 31. TOP CANDIDATE VISUALIZATION
# ================================================================

top_candidates = (
    merged_score
    .head(12)
    .sort_values(
        "Causal_Echo_Score"
    )
)

plt.figure(
    figsize=(11,7)
)

plt.barh(

    top_candidates["feature"],

    top_candidates[
        "Causal_Echo_Score"
    ]
)

plt.title(
    "Top Causal-Echo Candidates"
)

plt.xlabel(
    "Causal-Echo Score"
)

plt.tight_layout()
plt.show()

# ================================================================
# 32. ROBUSTNESS TEST — FEATURE NOISE
# ================================================================

noise_levels = [
    0.00,
    0.05,
    0.10,
    0.20,
    0.30
]

robustness_records = []

for noise in noise_levels:

    X_noisy = (
        X_test
        .copy()
    )

    noise_array = np.random.normal(
        0,
        noise,
        X_noisy.shape
    )

    X_noisy = (
        X_noisy +
        noise_array
    )

    noisy_prediction = (
        xgb.predict(
            X_noisy
        )
    )

    noisy_r2 = r2_score(
        y_test,
        noisy_prediction
    )

    noisy_mae = mean_absolute_error(
        y_test,
        noisy_prediction
    )

    robustness_records.append({

        "noise_level":
            noise,

        "R2":
            noisy_r2,

        "MAE":
            noisy_mae
    })

robustness_df = pd.DataFrame(
    robustness_records
)

print("\nNoise robustness")

display(
    robustness_df
)

# ================================================================
# 33. ROBUSTNESS VISUALIZATION
# ================================================================

plt.figure(
    figsize=(10,6)
)

plt.plot(

    robustness_df[
        "noise_level"
    ],

    robustness_df["R2"],

    marker="o",

    linewidth=3
)

plt.title(
    "Model Robustness Under Increasing Feature Noise"
)

plt.xlabel(
    "Noise level"
)

plt.ylabel(
    "R²"
)

plt.grid(
    alpha=0.3
)

plt.show()

# ================================================================
# 34. FEATURE ABLATION EXPERIMENT
# ================================================================
#
# Remove the top SHAP feature and retrain.
# This tests whether the model depends strongly on that feature.
#
# ================================================================

top_feature = (
    shap_df.iloc[0]
    ["feature"]
)

print(
    "\nTop SHAP feature:",
    top_feature
)

X_ablation = (
    X.drop(
        columns=[
            top_feature
        ]
    )
)

X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(

    X_ablation,

    y,

    test_size=0.20,

    random_state=42
)

xgb_ablation = XGBRegressor(

    n_estimators=250,

    max_depth=5,

    learning_rate=0.04,

    subsample=0.85,

    colsample_bytree=0.85,

    random_state=42,

    n_jobs=-1
)

xgb_ablation.fit(
    X_train_a,
    y_train_a
)

ablation_pred = (
    xgb_ablation.predict(
        X_test_a
    )
)

ablation_r2 = r2_score(
    y_test_a,
    ablation_pred
)

print(
    "Original XGBoost R²:",
    round(xgb_r2,4)
)

print(
    "After removing top feature:",
    round(ablation_r2,4)
)

print(
    "R² change:",
    round(
        ablation_r2 - xgb_r2,
        4
    )
)

# ================================================================
# 35. ROOT-CAUSE CANDIDATE ANALYSIS
# ================================================================

# Features that:
# 1. Change early
# 2. Have strong downstream network influence
# 3. Are important to ML
# 4. Show strong deviation

root_candidates = (
    merged_score
    .sort_values(
        "Causal_Echo_Score",
        ascending=False
    )
    .head(10)
)

print("\n")
print("="*90)
print("POTENTIAL EARLY-DISTURBANCE CANDIDATES")
print("="*90)

for i, (_, row) in enumerate(
    root_candidates.iterrows(),
    start=1
):

    print(
        f"\n{i}. {row['feature']}"
    )

    print(
        "Causal-Echo Score:",
        round(
            row[
                "Causal_Echo_Score"
            ],
            4
        )
    )

    print(
        "First change:",
        row[
            "first_change_time"
        ]
    )

    print(
        "Network betweenness:",
        round(
            row[
                "betweenness"
            ],
            4
        )
    )

    print(
        "SHAP importance:",
        round(
            row[
                "mean_abs_SHAP"
            ],
            4
        )
    )

# ================================================================
# 36. AUTOMATIC RESEARCH HYPOTHESES
# ================================================================

print("\n")
print("="*90)
print("AUTOMATICALLY GENERATED COMPUTATIONAL HYPOTHESES")
print("="*90)

for i, (_, row) in enumerate(
    root_candidates.head(8).iterrows(),
    start=1
):

    feature = row["feature"]

    print(
        f"\nHypothesis {i}"
    )

    print(
        f"{feature} appears as an early-changing "
        f"and computationally important variable."
    )

    print(
        "Potential interpretation:"
    )

    print(
        f"The temporal, network, and ML signals for "
        f"{feature} make it a candidate for further "
        f"investigation as part of the simulated "
        f"propagation cascade."
    )

    print(
        "Required validation:"
    )

    print(
        "Real longitudinal data + intervention/perturbation "
        "experiments + independent biological evidence."
    )

# ================================================================
# 37. SAVE RESEARCH OUTPUTS
# ================================================================

data.to_csv(
    "CausalEcho_temporal_biological_data.csv",
    index=False
)

summary.to_csv(
    "CausalEcho_feature_statistics.csv"
)

temporal_df.to_csv(
    "CausalEcho_temporal_precedence.csv",
    index=False
)

model_results.to_csv(
    "CausalEcho_model_comparison.csv",
    index=False
)

importance_df.to_csv(
    "CausalEcho_XGBoost_importance.csv",
    index=False
)

permutation_df.to_csv(
    "CausalEcho_permutation_importance.csv",
    index=False
)

shap_df.to_csv(
    "CausalEcho_SHAP_importance.csv",
    index=False
)

lag_df.to_csv(
    "CausalEcho_lagged_relationships.csv",
    index=False
)

edge_evidence_df.to_csv(
    "CausalEcho_discovered_edges.csv",
    index=False
)

network_df.to_csv(
    "CausalEcho_network_centrality.csv",
    index=False
)

merged_score.to_csv(
    "CausalEcho_Causal_Echo_Scores.csv",
    index=False
)

transitions_df.to_csv(
    "CausalEcho_state_transitions.csv",
    index=False
)

robustness_df.to_csv(
    "CausalEcho_robustness_analysis.csv",
    index=False
)

root_candidates.to_csv(
    "CausalEcho_root_candidates.csv",
    index=False
)

# ================================================================
# 38. FINAL RESEARCH DASHBOARD
# ================================================================

print("\n")
print("="*90)
print("CAUSAL-ECHO AI — FINAL RESEARCH DASHBOARD")
print("="*90)

print(
    "\nDataset:"
)

print(
    "Samples:",
    N_SAMPLES
)

print(
    "Temporal observations:",
    len(data)
)

print(
    "Biological variables:",
    N_BIO_FEATURES
)

print(
    "\nMachine Learning:"
)

print(
    "Random Forest R²:",
    round(rf_r2,4)
)

print(
    "XGBoost R²:",
    round(xgb_r2,4)
)

print(
    "\nCausal-structure recovery:"
)

print(
    "True edges:",
    len(true_edge_set)
)

print(
    "Discovered edges:",
    len(discovered_edges)
)

print(
    "Recovered true edges:",
    len(true_edges_found)
)

print(
    "Structural precision:",
    round(
        precision_discovery,
        4
    )
)

print(
    "Structural recall:",
    round(
        recall_discovery,
        4
    )
)

print(
    "\nTop Causal-Echo candidate:"
)

print(
    merged_score.iloc[0]["feature"]
)

print(
    "Score:",
    round(
        merged_score.iloc[0][
            "Causal_Echo_Score"
        ],
        4
    )
)

print(
    "\nTop SHAP feature:"
)

print(
    shap_df.iloc[0]["feature"]
)

print(
    "\nTop temporal early-change feature:"
)

print(
    temporal_df.iloc[0]["feature"]
)

print(
    "\nFiles generated:"
)

output_files = [

    "CausalEcho_temporal_biological_data.csv",
    "CausalEcho_feature_statistics.csv",
    "CausalEcho_temporal_precedence.csv",
    "CausalEcho_model_comparison.csv",
    "CausalEcho_XGBoost_importance.csv",
    "CausalEcho_permutation_importance.csv",
    "CausalEcho_SHAP_importance.csv",
    "CausalEcho_lagged_relationships.csv",
    "CausalEcho_discovered_edges.csv",
    "CausalEcho_network_centrality.csv",
    "CausalEcho_Causal_Echo_Scores.csv",
    "CausalEcho_state_transitions.csv",
    "CausalEcho_robustness_analysis.csv",
    "CausalEcho_root_candidates.csv"
]

for filename in output_files:

    print(
        "✓",
        filename
    )

# ================================================================
# 39. SCIENTIFIC LIMITATIONS
# ================================================================

print("\n")
print("="*90)
print("IMPORTANT SCIENTIFIC LIMITATIONS")
print("="*90)

print("""
1. This notebook uses SIMULATED biological-style data.

2. The hidden causal graph is known because it was used to
   generate the synthetic dataset.

3. The discovered relationships therefore demonstrate an
   algorithmic experiment, not a biological discovery.

4. Temporal precedence and lagged correlation do NOT establish
   biological causality by themselves.

5. The Causal-Echo Score is a computational prioritization
   heuristic, not a formal causal estimator.

6. SHAP measures model attribution, not biological causation.

7. Network centrality describes graph structure and does not
   automatically imply biological importance.

8. Real biological applications require carefully designed
   observational or intervention datasets.

9. Real-world validation should include:
   - longitudinal experiments
   - perturbation/intervention data
   - biological replicates
   - batch-effect correction
   - independent datasets
   - pathway annotation
   - literature evidence
   - experimental validation

10. The next research stage would be to replace the simulated
    system with real transcriptomic, proteomic, metabolomic,
    single-cell, or perturbation datasets.
""")

# ================================================================
# 40. COMPLETION
# ================================================================

print("\n")
print("="*90)
print("CAUSAL-ECHO AI COMPLETED SUCCESSFULLY")
print("="*90)

print("""
Research pipeline completed:

Synthetic Biology
        ↓
Temporal Perturbation
        ↓
Statistical Analysis
        ↓
Random Forest
        ↓
XGBoost
        ↓
Feature Importance
        ↓
SHAP
        ↓
Anomaly Detection
        ↓
Hidden State Discovery
        ↓
Temporal Precedence
        ↓
Lagged Relationships
        ↓
Propagation Network
        ↓
Network Centrality
        ↓
Structure Recovery
        ↓
Robustness Testing
        ↓
Feature Ablation
        ↓
Causal-Echo Score
        ↓
Root-Cause Candidates
        ↓
Computational Hypotheses

CAUSAL-ECHO AI: COMPLETE
""")
```
